```python
file = "selected_papers_post_gpt.json"
subdir = "/post_pdf/"
subdir_txt = "/post_txt/"
limit = 1000
```

```python
# Read papers
import json

with open(file, "r") as f:
    data = json.load(f)
```

```python
len(data)
```

```python
import requests
import time
from pathlib import Path

i = 0
while i < limit:
    # Get link and create filename from link
    link = data[i]["link"]
    file = Path("./papers" + subdir + link.split("/pdf/")[1] + ".pdf")
    
    # Check if file exists
    if not file.exists():
        # download paper
        download = requests.get(link)
        if download.headers["Content-Type"] == "application/pdf":
            with open(file, "wb") as f:
                f.write(download.content)
                print(f"Stored {file} [{i+1}/{limit}]")
        # Wait 1 s
        time.sleep(1)
        
        # Check if file exists now
        if not file.exists():
            limit += 1 # File can't be download, increase limit by one
    else:
        print(f"File exists [{i+1}/{limit}]")
        
    i += 1
```

```python
# Convert to txt
import subprocess
for i in range(limit):
    print(f"{i}/{limit}")
    # Get link and create filename from link
    link = data[i]["link"]    
    file_pdf = Path("./papers" + subdir + link.split("/pdf/")[1] + ".pdf")
    file_txt = Path("./txt" + subdir_txt + link.split("/pdf/")[1] + ".txt")
    if not file_txt.exists():
        subprocess.run(["java", "-jar", "../PDFMining/build/libs/PDFExtract-1.0-SNAPSHOT.jar", f"--pdf=./{file_pdf}", f"--txt=./{file_txt}"])
```
