```python
import arxiv
client = arxiv.Client()
```

```python
search = arxiv.Search(
    query="cat:cs.* AND submittedDate:[20150101 TO 20220101]",
    max_results = 1500,
    sort_by = arxiv.SortCriterion.SubmittedDate
)
```

```python
results = client.results(search)
```

```python
result_list = list(results)
```

```python
arxiv_list = []
for entry in result_list:
    if entry.published.year < 2022:
        arxiv_list.append({
            "title": entry.title,
            "year": entry.published.year,
            "link": str(entry.links[1])
        })
```

```python
arxiv_list
```

```python
len(arxiv_list)
```

```python
import json

with open("./selected_papers_pre_gpt.json", "w") as f:
    json.dump(arxiv_list, f)
```
