```python
from langchain_ollama import ChatOllama

SYSTEM_PROMPT = (
    "Create a scientific paper for the given title. It should have a length of 12 DIN a4 pages. Don't ask the user anything. Always write full sentences and include a literature review."
)

model = ChatOllama(
    base_url="http://localhost:11434",
    model="gemma3:27b",
)
```

```python
messages = [
    ("system", SYSTEM_PROMPT),
    ("human", "A Leaner and Faster Web: How CBOR Can Improve Dynamic Content Encoding in JSON and DNS over HTTPS")
]
```

```python
response = model.invoke(messages)
```
