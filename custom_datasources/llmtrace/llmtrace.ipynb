Converts https://huggingface.co/datasets/iitolstykh/LLMTrace_classification from JSONL to csv


```python
import json

texts = []
generated = []

with open("./train.jsonl", "r") as f:
    for line in f:
        line = line.strip()
        line = json.loads(line)
        if line["lang"] == "eng":
            texts.append(line["text"])
            if line["label"] == "human":
                generated.append(0)
            else:
                generated.append(1)

with open("./test.jsonl", "r") as f:
    for line in f:
        line = line.strip()
        line = json.loads(line)
        if line["lang"] == "eng":
            texts.append(line["text"])
            if line["label"] == "human":
                generated.append(0)
            else:
                generated.append(1)

with open("./valid.jsonl", "r") as f:
    for line in f:
        line = line.strip()
        line = json.loads(line)
        if line["lang"] == "eng":
            texts.append(line["text"])
            if line["label"] == "human":
                generated.append(0)
            else:
                generated.append(1)
```

```python
# Store data
import pandas as pd

df = pd.DataFrame({
    "text": texts,
    "generated": generated
})
```
