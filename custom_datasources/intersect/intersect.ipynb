```python
# Get download links
links = [link.replace("/view/", "/download/") for link in links]
```

```python
links
```

```python
# Download the papers
import requests
import time

for i in range(len(links)):
    download = requests.get(links[i])
    if download.headers["Content-Type"] == "application/pdf":
        filename = f"./pdf/{i}.pdf"
        with open(filename, "wb") as f:
            f.write(download.content)
            print(f"Stored {filename}")

    time.sleep(1)
```

```python
import subprocess

for i in range(len(links)):
    subprocess.run(["java", "-jar", "../../PDFMining/build/libs/PDFExtract-1.0-SNAPSHOT.jar", f"--pdf=./pdf/{i}.pdf", f"--txt=./txt/{i}.txt"])
```

```python
# Read txt files
import os
files = []
for (dirpath, dirnames, filenames) in os.walk("./txt/"):
    files.extend(filenames)

files
```

```python
# Read all files
documents = []

for file in files:
    with open(f"./txt/{file}", "r") as f:
        lines = []
        for line in f:
            line = line.strip()
            if line.lower().startswith("intersect"):
                continue
            if line.lower().startswith("references"):
                break

            if len(line) > 10:
                lines.append(line)

        documents.append(lines)
        
```

```python
documents
```

```python
texts = [item for sublist in documents for item in sublist]
```

```python
texts
```

```python
len(texts)
```

```python
# Concat texts
import tqdm

whole_text = ""
for i in tqdm.tqdm(range(len(texts))):
    whole_text += texts[i] + " "
```

```python
import nltk
from nltk.tokenize.punkt import PunktSentenceTokenizer, PunktParameters

nltk.download('punkt_tab')
params = PunktParameters()
params.sent_end_chars=('.',)
params.abbrev_types = {'e.g', 'i.e', 'e.g.', 'i.e.'} 
# Normalieze  'e.g.', 'et al.'

whole_text = whole_text.replace("e. g.", "_e_g_")
whole_text = whole_text.replace("et al.", "_et_al_")
tokenizer = PunktSentenceTokenizer(params)
sentences = tokenizer.tokenize(whole_text)

# 'unnormalize'
sentences = [s.replace("_e_g_", "e.g.") for s in sentences]
sentences = [s.replace("_et_al_", "et al.") for s in sentences]
```

```python
# Store
import pandas as pd
df = pd.DataFrame({
    "text": sentences,
    "generated": 0
})
```
