Read ChatGPT responses into CSV file

```python
lines = []

with open("batchoutput.jsonl", "r") as f:
    for line in f:
        lines.append(line.strip())
```

```python
import json

numbers = []
models = []
texts = []

for line in lines:
    line = json.loads(line)
    models.append(line["response"]["body"]["model"])
    texts.append(line["response"]["body"]["choices"][0]["message"]["content"])
```

```python
# Store data
import pandas as pd

df = pd.DataFrame({
    "model": models,
    "text": texts,
})
```

```python
df.to_csv("batchoutput.csv", index=False)
```
