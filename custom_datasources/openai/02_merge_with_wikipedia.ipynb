Merge data from OpenAI with Wikipedia data

```python
# Load wikipedia data
import pickle
with open("../wikipedia/selected_data.bin", "rb") as f:
    wikipedia = pickle.load(f)
```

```python
wikipedia[0]
```

```python
# Load GPT data
import pandas as pd
df = pd.read_csv("./batchoutput.csv")
```

```python
df.head()
```

```python
# Rename text to gpt_text
df.rename(columns={"text": "gpt_text"}, inplace=True)
```

```python
df.head()
```

```python
# Collect title and human texts
titles = []
human_texts = []

for item in wikipedia:
    titles.append(item["title"])
    human_texts.append(item["text"])
```

```python
# Insert data
df.insert(0, "title", titles)
df.insert(1, "human_text", human_texts)
```

```python
df.head()
```

```python
df.to_csv("../../data/datasets/wikipedia_gpt.csv", index=False)
```
