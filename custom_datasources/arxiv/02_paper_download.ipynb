```python
# Read papers
import json

with open("selected_papers.json", "r") as f:
    data = json.load(f)
```

```python
import requests
import time

for i in range(len(data)):
    # Get link and create filename from link
    link = data[i]["link"]
    filename = link.split("/pdf/")[1] + ".pdf"

    # download paper
    download = requests.get(link)
    if download.headers["Content-Type"] == "application/pdf":
        filename = f"./papers/{filename}"
        with open(filename, "wb") as f:
            f.write(download.content)
            print(f"Stored {filename}")

    # Wait 1 s
    time.sleep(1)
```

```python
# Convert to txt
import subprocess
for i in range(len(data)):
    # Get link and create filename from link
    link = data[i]["link"]
    filename = link.split("/pdf/")[1] + ".pdf"
    filename_txt = link.split("/pdf/")[1] + ".txt"
    subprocess.run(["java", "-jar", "../../PDFMining/build/libs/PDFExtract-1.0-SNAPSHOT.jar", f"--pdf=./papers/{filename}", f"--txt=./txt/{filename_txt}"])
```
