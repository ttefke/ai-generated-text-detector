```python
import pandas as pd

df = pd.read_csv("batchoutput.csv")
```

```python
df.head()
```

```python
texts = df["text"]
```

```python
non_ascii = dict()

for text in texts:
    for character in text:
        if ord(character) > 127:
            if character in non_ascii:
                non_ascii[character] += 1
            else:
                non_ascii[character] = 1
```

```python
# Unicode lookup table
unicode_lookup = {
    b'\\u2014': "Em dash",
    b'\\u2013': "En dash",
    b'\\u2011': "Non-breaking hyphen",
    b'\\u2019': "Right single quotation mark",
    b'\\u201c': "Left double quotation mark",
    b'\\u201d': "Right double quotation mark",
}
```

```python
for key, value in non_ascii.items():
    if value >= 100:
        print(f"{key}: {value}, unicode: {key.encode("unicode_escape")}, {ord(key)}: {unicode_lookup[key.encode("unicode_escape")] if key.encode("unicode_escape") in unicode_lookup else ""}")
```

```python
characters = dict()
nr_of_characters = 0

for text in texts:
    for c in text:
        nr_of_characters += 1.0
        if c in characters:
            characters[c] += 1.0
        else:
            characters[c] = 1.0
```
